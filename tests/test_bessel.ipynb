```python
# このNoteは、Coulomb関数, Bessel関数の計算をmpmath, scipyで行い、
# Fortranで実装したcoulccライブラリの結果と比較するためのもの
# kfn, ell, eta, thetaの意味はtest_coulcc.f90と同じ

# kfn = 0 : Coulomb F & G
# kfn = 1 : spherical Bessel j and y
# kfn = 2 : Bessel J and Y
# kfn = 3 : modified Bessel I and K

import numpy as np
import scipy as sp
import mpmath as mp
mp.dps = 25 # set decimal places for mpmath


kfn = 0  # set type of function: 0,-1=Coulomb, 1=spherical Bessel, 2=Bessel, 3=modified Bessel
ell = 1  # set rank of function
eta = 1.0 + 0.0j  # set Sommerfeld parameter (used only for Coulomb functions, kfn=0, -1)
theta = 0.8  # set phase of complex argument xx = exp(i*theta)*x
# set x from 0.00 to 5.00 with step 0.01
x = np.linspace(0.00, 5.00, 21)
xx = np.exp(theta*1j)*x


# compute the spherical Bessel function of the first kind of order 0
# if  KFN  =  0,-1  complex Coulomb functions are returned   F & G    C
if kfn == 0: # kfn = 0     Coulomb          F & G
  print("# kfn=",kfn," ell=",ell)
  print("# eta1=(Re,Im) ", eta.real, eta.imag)
  print("# theta= ", theta)
  print("#  x         Re(xx)    Im(xx)   F(x)             G(x)            Re(F(xx))        Im(F(xx))        Re(G(xx))        Im(G(xx))")
  for x_ in x:
      xx = np.exp(theta*1j) * x_
      F = float(mp.coulombf(ell, eta, x_))
      G = float(mp.coulombg(ell, eta, x_))
      CF = complex(mp.coulombf(ell, eta, xx))
      CG = complex(mp.coulombg(ell, eta, xx))
      print(f"{x_.real:9.4f} {xx.real:9.4f} {xx.imag:9.4f} {F:16.8e} {G:16.8e}"
            f"{CF.real:16.8e} {CF.imag:16.8e} {CG.real:16.8e} {CG.imag:16.8e}")
else:
  if kfn == 1:  # kfn = 1  spherical Bessel   j & y
    y0 = sp.special.spherical_jn(ell, x)
    y1 = sp.special.spherical_yn(ell, x)
    y2 = sp.special.spherical_jn(ell, xx) # kfn = 1, F
    y3 = sp.special.spherical_yn(ell, xx) # kfn = 1, G

  if kfn == 2:  # kfn = 2   Bessel J & Y
    y0 = sp.special.jv(ell, x)
    y1 = sp.special.yv(ell, x)
    y2 = sp.special.jv(ell, xx)
    y3 = sp.special.yv(ell, xx)

  if kfn == 3:  # kfn = 3   modified Bessel I & K
    y0 = sp.special.iv(ell, x)
    y1 = sp.special.kv(ell, x)
    y2 = sp.special.iv(ell, xx)
    y3 = sp.special.kv(ell, xx)

  # print results
  print("# kfn=",kfn,", ell=",ell)
  print("# eta1=(Re,Im) ", eta.real, eta.imag)
  print("# theta= ", theta)
  print("#  x         Re(xx)    Im(xx)   F(x)             G(x)            Re(F(xx))        Im(F(xx))        Re(G(xx))        Im(G(xx))")
  for i in range(len(xx)):
      print(f"{x[i].real:9.4f} {xx[i].real:9.4f} {xx[i].imag:9.4f} {y0[i]:16.8e} {y1[i]:16.8e}"
            f"{y2[i].real:16.8e} {y2[i].imag:16.8e} {y3[i].real:16.8e} {y3[i].imag:16.8e}")

```
